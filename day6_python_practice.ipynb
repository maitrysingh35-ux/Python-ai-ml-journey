```python
#A list is a collection of multiple items stored in a single variable. # mutable , separated by commas.
subjects = ["Python", "Java", "AI"]
print(subjects[0])
```

```python
#creating list with different data types.
student = ["Maitri", 19, 8.8, True]
print(student)
```

```python
#Different ways to create a list.
numbers = [5, 10, 15, 20, 25]
print(numbers[-1])
```

```python
#list slicing - list_name[start:end:step]
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[1:4])
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[::2])
numbers = [10, 20, 30, 40, 50]
print(numbers[::-1])
```

```python
# Updating List Elements
subjects = ["Python", "Java", "AI"]
subjects[1] = "Machine Learning"
print(subjects)

numbers = [10, 20, 30, 40]
numbers[2] = 100

print(numbers)
```

```python
subjects = ["Python", "Java", "C++", "AI"]
print(subjects)
print(subjects[0])
print(subjects[3])
subjects[1] = "Machine Learning"
print(subjects)
```

```python
# Adding elements to a List
# We'll learn three important methods:
# 1. append() – adds one element at the end.
subjects = ["Python", "Java", "AI"]
subjects.append("Machine Learning")
print(subjects)

# 2. insert() – adds an element at a specific position.
fruits = ["Apple", "Mango"]
fruits.insert(1, "Banana")
print(fruits)

#3. extend() – adds multiple elements.
fruits = ["Apple", "Mango"]
more_fruits = ["Banana", "Orange"]
fruits.extend(more_fruits)
print(fruits)

numbers = [10, 20]
numbers.extend([30, 40])
print(numbers)
```

```python
num = [10,20,30]
num.append(40)
print(num)

num.insert(1,15)
print(num)

num.extend([50,60])
print(num)
```

```python
# Removing elements from a List -
# We'll learn four methods:
# 1. remove() - It removes an element by its value, not its index.
fruits = ["Apple", "Mango", "Banana", "Orange"]
fruits.remove("Mango")
print(fruits)

# 2. pop() - pop() removes an element using its index. or pop() without an index removes the last element from the list.
fruits = ["Apple", "Mango", "Banana", "Orange"]
fruits.pop(1)
print(fruits)

numbers = [10, 20, 30, 40]
x = numbers.pop(2)
print(x)
print(numbers)

#3. del - del removes an element using its index. It can also delete an entire list.
numbers = [10, 20, 30, 40]
del numbers[1]
print(numbers)

numbers = [10, 20, 30, 40, 50]
del numbers[1:3]
print(numbers)

# 4. clear() - The clear() method removes all elements from a list, but the list itself still exists.
subjects = ["Python", "Java", "AI"]

subjects.clear()

print(subjects)
```

```python
subjects = ["Python", "Java", "C++", "AI", "DBMS"]
subjects.remove("Java")
print(subjects)

subjects.pop()
print(subjects)

del subjects [1]
print(subjects)

subjects.clear()
print(subjects)
```

```python
# Finding elements in a List
# We'll learn:
# 1.index() - The index() method returns the position of an element in a list.
subjects = ["Python", "Java", "AI", "DBMS"]
print(subjects.index("AI"))

# 2.count() - The count() method tells us how many times an element appears in a list.
numbers = [10, 20, 30, 20, 40, 20]
print(numbers.count(20))
# 3. - in and not in - These are membership operators. They check whether an element exists in a list.
# in 
subjects = ["Python", "Java", "AI"]
print("Python" in subjects)
# not in 
subjects = ["Python", "Java", "AI"]
print("C++" not in subjects)
# 5. copy - The copy() method creates a separate copy of an existing list.
subjects = ["Python", "Java", "AI"]

new_subjects = subjects.copy()

print(subjects)
print(new_subjects)
```

```python
# sorting a list 
# 1. sort() - sorting the original list
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)
# 2. sort(reverse=True) – descending order
numbers = [40, 10, 30, 20]
numbers.sort(reverse=True)
print(numbers)
#4. sorted - it does not change the original list , it return a new sorted list.
numbers = [40, 10, 30, 20]
new_numbers = sorted(numbers)
print(new_numbers)
print(numbers)
#3. reverse() - reverse th exist order.
numbers = [40, 10, 30, 20]
numbers.reverse()
print(numbers)
```

```python
# list Operations
# We'll cover:
# 1. Concatenation (+)
a = [10, 20, 30]
b = [40, 50]
c = a + b
print(c)

# 2. Repetition (*) - The * operator repeats the elements of a list.
numbers = [1, 2, 3]
print(numbers * 2)

# 3. Length (len()) - The len() function returns the total number of elements in a list., count element not index numbers.
subjects = ["Python", "Java", "AI", "DBMS"]
print(len(subjects))

# 4. Maximum (max())
numbers = [10, 50, 20, 80, 30]
print(max(numbers))

# 5. Minimum (min())
numbers = [10, 50, 20, 80, 30]
print(min(numbers))

# 6. Sum (sum()) - The sum() function adds all numerical elements in a list.
numbers = [10, 20, 30, 40]
print(sum(numbers))
```

```python
# Nested Lists 🐍 - A nested list is simply a list inside another list.
students = [
    ["Amit", 85],
    ["Riya", 92],
    ["Neha", 78]
]

print(students)

print(students[0][0]) # first inner list 
print(students[0][1]) # first element inside that list
```
