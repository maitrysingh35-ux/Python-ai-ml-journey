```python
#String methods
#1.Lower()
name = "MAITRI"
print(name.lower())
#2.UPPER()
name = "maitri"
print(name.upper())
#3. title() - Converts the first letter of every word into uppercase.
text = "i love python programming"
print(text.title())
#4. Capitalize() - Converts only the first character of the string into uppercase and the remaining characters into lowercase.
text = "hELLO PYTHON"
print(text.capitalize())
```

```python
text = "welcome to PYTHON programming"
print(text.lower())
print(text.upper())
print(text.title())
print(text.capitalize())
```

```python
#swapcase()
text = "Hello Python"
print(text.swapcase())
```

```python
text = "I Love PYTHON"
print(text.swapcase())
```

```python
#strip() - removes extra spaces from the beginning and end of a string.
text = "   Hello Python   "
print(text.strip())
#lstrip() - Removes spaces from the left side only.
text = "  Hello "
print(text.lstrip())
#rstrip()- Removes spaces from the right side only.
text = "   Hello   "
print(text.rstrip())
```

```python
text = "  Python Programming  "
print(text.strip())
print(text.lstrip())
print(text.rstrip())
```

```python
#replace() - replace() is used to replace one character or word with another.
#string.replace(old, new)
text = "I love Java"
print(text.replace("Java", "Python"))
```

```python
text = "I love cats"
print(text.replace("cats","dogs"))
#occurence replace 
text = "Python Python Python"
print(text.replace("Python", "Java", 1))
text = "apple apple apple"
print(text.replace("apple", "mango", 2))
```

```python
#split()- breaks a string into smaller parts and returns them in a list.
text = "I love Python"
print(text.split())
text = "apple,banana,mango"
print(text.split(","))
```

```python
#join() - joins multiple strings together.
words = ["I", "love", "Python"]
result = " ".join(words)
print(result)
```

```python
words = ["apple", "banana", "mango"]
answer = ",".join(words)
print(answer)
```

```python
#find() - searches for a character or word inside a string and returns its index position
text = "I love Python"
print(text.find("Python"))
```

```python
text = "I love Python programming"
print(text.find("Python"))
print(text.find("love"))
```

```python
#count()- count() tells us how many times a character or word appears in a string.
text = "banana"
print(text.count("a"))
word = "MISSISSIPPI"
print(word.count("S"))
```

```python
text = "Python is easy. Python is powerful. Python is fun."
print(text.count("Python"))
```

```python
#startswith() and endswith()
#These methods check whether a string starts or ends with a particular character or word.
#1. startswith()
text = "Python Programming"
print(text.startswith("Python"))
#2. endswith()
text = "Python Programming"
print(text.endswith("Programming"))
```

```python
#isalpha() – checks whether all characters are alphabets.
text = "Python"
print(text.isalpha())

#isdigit() – checks whether all characters are digits.
text = "12345"
print(text.isdigit())

#isalnum() – checks whether all characters are alphabets or digits (or a combination of both).
text = "Python123"
print(text.isalnum())
```

```python
#String Concatenation - Concatenation means joining two or more strings together.
# We use the + operator.
first = "Hello"
second = "Python"

result = first + second
print(result)

first = "Hello"
second = "Python"

result = first + " " + second
print(result)

name = "Maitri"
course = "B.Tech"
college = "KPGU"

print(name + " " + course + " " + college)

age = 19
print("My age is " + str(age))
```

```python
first = "Generative"
second = "AI"
third = "Engineer"
print("I want to become a " + first + " " + second + " " + third )
```

```python
# String repetition means repeating a string multiple times.
# We use the * operator.
word = "Python "
print(word * 3)
text = "Hello\n"
print(text * 3)
```

```python
text = " AI "
print(text *  4)
```

```python
#Next: Escape Sequences - Escape sequences are special characters used inside strings to control formatting.
#1. \n - new line
print("Hello\nPython")
# 2. \t	- Tab space
print("Hello\tPython")
# 3. \\ - 	Prints a backslash
print("C:\\Python\\Files")
#\'	Prints a single quote
print("C:\'Python\'Files")
#\"	Prints a double quote
print("C:\"Python\"Files")
```

```python
#Multi-line Strings - Python allows us to write a string across multiple lines using triple quotes (""" """ or ''' ''').
message = """Hello!
Welcome to Python.
Let's learn programming."""

print(message)
```

```python
message = """welcome to KPGU!
I am learning Python.
My goal is to become AI Engineer."""

print(message)
```
