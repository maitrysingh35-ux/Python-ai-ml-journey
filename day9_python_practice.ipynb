```python
# Accessing Dictionary Elements (Keys and Values).
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
# Accessing values using keys: dict[key]
print(student["name"])
print(student["age"])

#Accessing values using get()
print(student.get("name"))
print(student.get("course"))

#Difference between [] and get()
print(student.get("city"))
print(student.get("city", "Not Found"))
print(student["city"])
```

```python
#Accessing keys and values using keys() and values()
# using keys()
student = {
    "name": "Riya",
    "age": 20,
    "course": "CSE"
}

print(student.keys())
print(student.values())
```

```python
# Adding New Key-Value Pairs
# 1. Using square brackets []
student = {
    "name": "Maitri",
    "age": 19
}

student["city"] = "Vadodara"
print(student)

# 2. using update() method
student = {
    "name": "Maitri",
    "age": 19
}
student.update({"city": "Vadodara", "course": "CSE"})
print(student)
```

```python
# Updating Existing Values
# 1. 1. Updating using square brackets []
student = {
    "name": "Amit",
    "age": 19,
    "course": "Java"
}

student["age"] = 20
student["course"] = "Python"

print(student)
# 2. use update()
student = {
    "name": "Riya",
    "age": 20,
    "course": "Java"
}

student.update({"age": 21, "course": "Python"})

print(student)
```

```python
# Removing Elements from Dictionaries.
# 1. pop() - Removes a specific key and returns its value.
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
student.pop("age")
print(student)

# 2. popitem() -  Removes and returns the last inserted key-value pair.
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
student.popitem()
print(student)

# 3. del() - remove a specific key-value pair.
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
del student["age"]
print(student)

# 4.clear () - Removes all elements from the dictionary but keeps the dictionary itself empty.
student = {
    "name": "Maitri",
    "age": 19
}
student.clear()
print(student)
```

```python
# DICTIONARY METHODS
# 1. keys() - return all keys.
student = {"name": "Riya", "age": 20}
print(student.keys())
# 2. values() - return all values.
print(student.values())
# 3. items() - returns all key-value pairs as tuples.
print(student.items())
# 4. get() - Returns the value associated with a key.
print(student.get("name"))
student.get("marks", 0)
# 5. update() - Adds new pairs or update existing values.
student.update({"age": 21})
print(student)
```

```python
# Traversing Dictionaries Using Loops.
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
# 1. looping through keys  - When we use a for loop directly on a dictionary, Python gives us its keys.
for key in student:
    print(i)
for i in student.keys():
    print(i)

# 2. looping through values - use values()
for value in student.values():
    print(value)
# 3.  looping through key-value pairs using items() 
for key, value in student.items():
    print(key, value)
```

```python
# membership operators  - check keys by default , not values.
# 1. in operator 
student = {
    "name": "Maitri",
    "age": 19,
    "course": "CSE"
}
print("name" in student)
print("city" in student)

# 2. not in operator 
print("city" not in student)
print("age" not in student)

# 3. checking whether a value exists.
print("Maitri" in student.values())
print(25 in student.values())
```

```python
# Nested dictionary 
students = {
    "student1": {
        "name": "Amit",
        "age": 20
    },
    "student2": {
        "name": "Riya",
        "age": 19
    }
}

print(students)

# 2. Accessing Inner dictionary values.
print(students["student1"]["name"])
print(students["student2"]["age"])

# 3. Updating Inner dictionary values
students["student1"]["age"] = 21
print(students["student1"])

# 4. Adding elements to inner dictionaries.
students["student1"]["city"] = "Ahmedabad"
print(students["student1"])

# 5. removing elements from inner dictionaries.
students["student1"].pop("city") 
print(students["student1"])
```

```python
# Dictionary comprehension
# syntax - {key: value for item in iterable}
squares = {i: i*i for i in range(1, 6)}
print(squares)
even = {i: i*i for i in range(1, 11) if i % 2 == 0}
print(even)

# normal method 
squares = {}
for i in range(1, 6):
    squares[i] = i * i
print(squares)
```

```python
# Type Conversion
# 1. Dictionary to list
student = {"name": "Amit", "age": 20, "marks": 85}
print(list(student))
print(list(student.values()))
print(list(student.items()))

# 2. Dictionary to Tuple
data = {"a": 10, "b": 20}
print(tuple(data))
print(tuple(data.values()))
print(tuple(data.items()))

# 3. Dictionary to set 
data = {"num1": 10, "num2": 20}
print(set(data))
print(set(data.values()))
print(set(data.items()))

# 4. converting other data types
data = [("name", "Amit"), ("age", 20)]
print(dict(data))
```
