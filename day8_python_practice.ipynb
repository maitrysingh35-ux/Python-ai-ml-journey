```python
# Properties of Sets.
# 1. Unordered - Sets do not maintain a fixed order of elements.
numbers = {10, 20, 30, 40}
print(numbers)

# 2. Unique elements - Sets do not allow duplicate values.
numbers = {10, 20, 10, 30, 20}
print(numbers)

# 3. Mutable - Mutable means we can change a set after creating it. , we can add or remove elements.
fruits = {"Apple", "Mango"}
fruits.add("Banana")
print(fruits)

# 4.No indexing or slicing - Sets do not support indexing or slicing.
numbers = {10, 20, 30, 40}
print(numbers[0])
```

```python
# Accessing and Traversing Sets
# 1. 1. Using a for loop 
fruits = {"Apple", "Mango", "Banana"}

for fruit in fruits:
    print(fruit)

# 2. Traversing a set - Traversing simply means visiting each element one by one.
numbers = {10, 20, 30, 40}

for num in numbers:
    print(num * 1)
```

```python
# Adding Elements to Sets
# 1. add() – Adds one element to a set.
fruits = {"Apple", "Mango", "Banana"}
fruits.add("Orange")
print(fruits)

# 2. update() – Adds multiple elements to a set.
fruits = {"Apple", "Mango"}
fruits.update(("Banana", "Orange", "Grapes"))
print(fruits)
```

```python
subjects = {"Python", "Java", "C++"}
subjects.add("AI")
subjects.update(["DBMS", "CN"])
print(subjects)
print(len(subjects))

languages = {"Python", "Java"}
languages.add("C++")
languages.update(("JavaScript", "SQL"))
print(languages)
print(len(languages))
```

```python
# Removing Elements from Sets
# 1.remove() - remove a specific element from a set
fruits = {"Apple", "Mango", "Banana"}
fruits.remove("Mango")
print(fruits)

# 2. discard() - method also remove specific element.
fruits = {"Apple", "Mango", "Banana"}
fruits.discard("Mango")
print(fruits)

# 3. pop() - method removes and returns a random/arbitrary element from a set.
numbers = {10, 20, 30, 40}

removed = numbers.pop()

print("Removed:", removed)
print("Remaining:", numbers)

# 4.clear() - method remove all elements from a set.
numbers = {10, 20, 30}

numbers.clear()

print(numbers)
```

```python
numbers = {10, 20, 30, 40, 50}
numbers.remove(30)
numbers.discard(100)
print(numbers)
print(len(numbers))

subjects = {"Python", "Java", "C++", "DBMS"}
subjects.discard("Java")
subjects.pop()
print(subjects)
print(len(subjects))

languages = {"Python", "Java", "C++"}
languages.remove('Java')
languages.clear()
print(languages)
print(len(languages))
```

```python
# set operation
# 1. union(|) - combine all unique elements from both sets.
# 2. Intersection(&) - gives only the common elements present in both sets.
# 3. Difference (-) -  gives elements that exist in the first set but not in the second set.
# 4. Symmetric Difference(^) -  It gives elements that are in either set, but not in both.
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
print(A|B)
print(A&B)
print(A-B)
print(B-A)
print(A^B)
```

```python
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
print(A|B)
print(A&B)
print(A-B)
print(A^B)
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
```

```python
# Set Methods
# 1. Union() - Combines all unique elements from both sets.2. intersection 3. difference 4. symmetric_difference
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
```

```python
# MEMBERSHIP OPERATORS 
# 1.IN OPERATOR - return true if the element exsist in set.
fruits = {"Apple", "Mango", "Banana"}

print("Apple" in fruits)
print("Orange" in fruits)

# 2. NOT IN - return true if the elements does not exsist in set
fruits = {"Apple", "Mango", "Banana"}

print("Orange" not in fruits)
print("Apple" not in fruits)
```

```python
# Set Comparison Methods
A = {1, 2}
B = {1, 2, 3, 4}
C = {5, 6}

#  1.  issubset() - Checks whether all elements of one set are present in another set.
print(A.issubset(B))
# 2. issuperset() - Checks whether a set contains all elements of another set.
print(B.issuperset(A))
# 3. isdisjoint() - Checks whether two sets have no common elements.
print(A.isdisjoint(C))
```

```python
# frozen set() -  immutable 
# 1. Creating a frozenset 
numbers = frozenset([10, 20, 30, 40])

print(numbers)
print(type(numbers))

# Can we add or remove elements? no 
numbers = frozenset([10, 20, 30])
numbers.add(40)

# What operations are allowed? Union - Intersection,Difference,Symmetric difference,Membership checking (in, not in)
```

```python
# Set Comprehension -Set comprehension is a short way to create a set using a single line of code instead of writing a full for loop. 
# syntax {expression for item in iterable}
squares = {i ** 2 for i in range(1, 6)}
print(squares)

even = {i for i in range(1, 11) if i % 2 == 0}
print(even)

squares = set()
for i in range(1, 6):
    squares.add(i ** 2)

print(squares)
```

```python
# Type Conversion – List ↔ Set ↔ Tuple
# 1. list - set  -  use set()
numbers = [10, 20, 30, 20, 40, 10]

new_numbers = set(numbers)

print(new_numbers)

# 2. set - list   - use list()
numbers = {10, 20, 30, 40}

new_numbers = list(numbers)

print(new_numbers)

# 3. tuple - set  - use set()
numbers = (10, 20, 30, 20, 40)

new_numbers = set(numbers)

print(new_numbers)

# 4. set - tuple  - use tuple()
numbers = {10, 20, 30, 40}

new_numbers = tuple(numbers)

print(new_numbers)
```
