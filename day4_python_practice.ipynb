```python
#args - allows function to accept any number of positional arguments(conventional name *numbers)
def add(*numbers):
    print(numbers)

add(10, 20, 30, 40)
```

```python
def add(*numbers):
    total = 0

    for i in numbers:
        total = total + i

    print(total)

add(10, 20, 30)
```

```python
def multiply(*value):
    result = 1
    for i in value:
        result = result * i
    print(result)
multiply(2,3,4)
```

```python
def largest(*value):
    maximum = 0
    for num in value:
        if num > maximum:
            maximum = num
    print(maximum)
largest(12,45,23,67,34)
```

```python
def count_even(*num):
    count = 0
    for i in num:
        if i % 2 == 0:
            count = count + 1
    print(count)

count_even(10,15,20,0,4,23,30,7,8)
```

```python
#kwargs - allows a function to accept multiple keyword arguments. it store all those arguments into a dictionary
def student(**details):
    print(details)

student(name="Maitri", age=19, college="KPGU")
```

```python
def student(**info):
    print(info["name"])
    print(info["college"])

student(name="Maitri", college="KPGU")
```

```python
def student_details(**value):
    print(value)
    print(value["name"])
    print(value["age"])

student_details(name = "Maitri", age = 19 , college = "KPGU")
    
```

```python
def guessing_name():
    secret_number = 11
    guess = 0
    while guess != secret_number:
        guess = int(input("Guess the number: "))
        if guess > secret_number:
            print("Too high ")
        elif guess < secret_number:
            print("Too low")
        else:
            print("You WON !")
guessing_name()
```
