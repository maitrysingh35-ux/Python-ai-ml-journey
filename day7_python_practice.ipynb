Python Tuples
We'll cover:
1. Creating tuples
2. Tuple indexing and negative indexing
3. Tuple slicing
4. Tuple immutability
5. Tuple methods: count() and index()
6. Tuple operations
7. Nested tuples
8. Converting between lists and tuples

```python
# A tuple is a collection of multiple elements stored in a single variable, just like a list. Tuples are immutable (not changeable).
# creating a tuple
fruits = ("Apple", "Mango", "Banana")

print(fruits)
print(type(fruits))

num = (10,20,30)
print(num)
print(type(num))

# Creating a Single-Element Tuple    
a = ("Apple",)
print(type(a))
```

```python
# Tuple Indexing 
fruits = ("Apple", "Mango", "Banana", "Orange")

print(fruits[0])
print(fruits[2])
print(fruits[-1])
```

```python
# Tuple slicing - tuple[start:stop:step]
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
```

```python
# Tuple Immutability - A tuple is immutable, which means we cannot change its elements after creating it.
subjects = ("Python", "Java", "AI")
subjects[1] = "C++"
```

```python
# Tuple methods — count() and index()
numbers = (10, 20, 10, 30, 10, 40)

print(numbers.count(10))
print(numbers.index(30))
```

```python
# Tuple Operations
# 1. Concatenation (+) - joins two tuples
a = (10, 20)
b = (30, 40)
print(a + b)

# 2.Repetition (*) - repeat a tuple
a = (1, 2)
print(a * 3)

# 3. membership (in , not in) - check whether an elemnet exists.
a = (10, 20, 30)

print(20 in a)
print(50 not in a)
```

```python
# Nested Tuples - A nested tuple means a tuple inside another tuple.
students = (
    ("Amit", 85),
    ("Riya", 92),
    ("Neha", 78)
)
print(students[1][0])
print(students[2][1])
```

```python
# Converting Lists and Tuples 
# 1. tuple () - convert a list into a tuple
fruits = ["Apple", "Mango", "Banana"]

result = tuple(fruits)
print(result)

# 2. List() - convert a tuple into a list
colors = ("Red", "Blue", "Green")

result = list(colors)
print(result)
```
